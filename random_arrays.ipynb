### 0. Создавайте функцию для записи списка в текстовый файл. Используйте CSV формат.

````python
import os

def to_file(arr: list, name: str = 'file') -> bool:
    '''
    Функция для записи списка в формате .csv
    :param arr: список, который будет записан в формате .csv
    :param name: имя файла без расширения, по умолчанию "file"
    :return: True при удачном выполнении кода, иначе False
    '''
    script_dir = script_dir = os.getcwd() 
    file_path = os.path.join(script_dir, f"{name}.csv")
    try:
        with open(f'{file_path}', 'w', encoding='utf-8') as f:
            for i in arr:
                f.write(f'{i}\n')
        return True
    except Exception as e:
        print(f'Ошибка: {e}')
        return False

to_file([1, 2, 3])
````

### 1. Создайте набор из случайных чисел без повторов $x_{1}, x_{2}, ... , x_{n}$

Создайте набор из нечетных чисел из случайных чисел без повторов $x_{1}, x_{2}, ... , x_{n}$. Интервал случайных значений *(a, b)*  и количество случайных чисел задаются пользователем. *n << (b - a)*

- **Создайте несколько реализаций:**

- **Сравните время работы этих подходов для:**
1. $n = 10^{3}$

Инициализация общих для всех функций данных:

````python
import numpy as np
from random import randint
import time

A = 0 #Нижняя граница диапазона чисел
B = 1_000_000_000 #Верхняя граница диапазона чисел
N = 1_000_00 #Количество чисел
rng = np.random.default_rng() #Инициализация генератора случ. чисел

#Декоратор для подсчета времени работы
def working_time(func):
    def wrapper(A, B, N):
        start = time.perf_counter()
        arr = func(A, B, N)
        end = time.perf_counter()
        print(f"{func.__name__} работает за  {end - start:.3f} сек.")
        return arr
    return wrapper
````

#### 1. Уникальный список из нечетных чисел при помощи list() + random.randint

````python
@working_time
def random_list(A: int, B: int, N: int) -> list:
    '''
    Функция для создания списка с уникальными нечетными числами
    :param A: Нижняя граница диапазона чисел
    :param B: Верхняя граница диапазона чисел
    :param N: Количество чисел
    :return: Список с уникальными нечетными числами
    '''

    arr = []
    while len(arr) < N:
        rnd = randint(A, B)
        if rnd not in arr and rnd % 2 == 1:
            arr.append(rnd)
    return arr
````

#### 2. Множество с нечетными числами

````python
@working_time
def random_set(A: int, B: int, N: int) -> set:
    '''
    Функция для создания множества с уникальными нечетными числами
    :param A: Нижняя граница диапазона чисел
    :param B: Верхняя граница диапазона чисел
    :param N: Количество чисел
    :return: Множество с уникальными нечетными числами
    '''

    arr = set()
    while len(arr) < N:
        rnd = randint(A, B)
        if rnd % 2 == 1:
            arr.add(rnd)
    return arr
````

#### 3. ndarray со случайными числами 

##### Актуальная версия через генератор случайных чисел $\color{#DC143C}{default\_rng()}$

1. $\color{#008B8B}{arange() + choice()}$

````python
@working_time
def random_arange_ndarray(A: int, B: int, N: int) -> np.ndarray:
    '''
    Функция для создания массива numpy.ndarray с уникальными нечетными числами
    :param A: Нижняя граница диапазона чисел
    :param B: Верхняя граница диапазона чисел
    :param N: Количество чисел
    :return: numpy.ndarray с уникальными нечетными числами
    '''

    # Необходимо, чтобы нижняя граница диапазона чисел была нечетной
    start = A if A % 2 == 1 else A + 1
    # Генерируем массив с упорядоченными детерминированными (!) нечетными числами
    odd = np.arange(start, B, 2)
    #Случайная выборка чисел из массива с нечетными числами
    arr = rng.choice(odd, size=N, replace=False)
    return arr
````

2. $\color{#008B8B}{choice() + формула}$

Чтобы не генерировать массив из $\frac{b - a}{2}$ элементов, случайно выберем *N* чисел в диапазоне *(a, b)*, а при помощи формулы гарантируем нечетность

Так как у нас целые числа, то $a_{i+1} = a_{i} + 2 \Rightarrow$ $a_{i+k} = a_{i} + 2k$

В данном случае k (в коде это массив arr) - индексы, которые случайно получены из **rng.choice()**, $a_{i}$ - минимальное нечетное число из *(a, b)*

````python
@working_time
def random_ndarray(A: int, B: int, N: int) -> np.ndarray:
    '''
    Функция для создания массива numpy.ndarray с уникальными нечетными числами
    :param A: Нижняя граница диапазона чисел
    :param B: Верхняя граница диапазона чисел
    :param N: Количество чисел
    :return: numpy.ndarray с уникальными нечетными числами
    '''

    # Необходимо, чтобы нижняя граница диапазона чисел была нечетной
    start = A if A % 2 == 1 else A + 1
    #Случайная выборка чисел из массива с нечетными числами
    arr = rng.choice((B - A) // 2, size=N, replace=False)
    return start + 2 * arr
````

##### Устаревшая версия np.random

1. $\color{#008B8B}{arange() + choice()}$

````python
@working_time
def random_arange_ndarray_old(A: int, B: int, N: int) -> np.ndarray:
    '''
    Функция для создания массива numpy.ndarray с уникальными нечетными числами
    :param A: Нижняя граница диапазона чисел
    :param B: Верхняя граница диапазона чисел
    :param N: Количество чисел
    :return: numpy.ndarray с уникальными нечетными числами
    '''

    # Необходимо, чтобы нижняя граница диапазона чисел была нечетной
    start = A if A % 2 == 1 else A + 1
    # Генерируем массив с упорядоченными детерминированными (!) нечетными числами
    odd = np.arange(start, B, 2)
    #Случайная выборка чисел из массива с нечетными числами
    arr = np.random.choice(odd, size=N, replace=False)
    return arr
````

2. $\color{#008B8B}{choice() + формула}$

````python
@working_time
def random_ndarray_old(A: int, B: int, N: int) -> np.ndarray:
    '''
    Функция для создания массива numpy.ndarray с уникальными нечетными числами
    :param A: Нижняя граница диапазона чисел
    :param B: Верхняя граница диапазона чисел
    :param N: Количество чисел
    :return: numpy.ndarray с уникальными нечетными числами
    '''

    # Необходимо, чтобы нижняя граница диапазона чисел была нечетной
    start = A if A % 2 == 1 else A + 1
    #Случайная выборка чисел из массива с нечетными числами
    arr = np.random.choice((B - A) // 2, size=N, replace=False)
    return start + 2 * arr
````

#### 4. Анализ производительности

1. ***random_list*** - худшая скорость. Пиковый расход - 3.84 МБ памяти, временная сложность O($N^{2}$) возникает из-за вложенных циклов: внешний while и внутренний, который реализован под капотом оператора in:
 ```python
if rnd not in arr
```
2. ***random_set*** - Пиковый расход - 8.40 МБ, временная сложность O($N$). Операция .add() использует хэш-таблицу $\Rightarrow$ проверка на уникальность и вставка выполняются в среднем за $O(1)$.
3. ***random_ndarray_old*** - Пиковый расход - 1908.12 МБ, временная сложность O($M$), где $M$ = $\frac{B - A}{2}$. В старой версии numpy `np.random.choice` под капотом делает перестановку в массиве из  $\frac{B - A}{2}$ элементов, а затем делает срез для получения N элементов
4. ***random_arange_ndarray_old*** - пиковый расход - 5722.88 МБ, временная сложность O($M$), где $M$ = $\frac{B - A}{2}$. Сначала создается массив на  $M$ нечетных чисел, а затем создается временный массив на такое же количество элементов для случайных перестановок (`np.random.choice`).
5. ***random_ndarray*** - пиковый расход - 2.29 МБ, временная сложность O($N$). `rng.choice()` генерирует ровно $N$ случайных индексов за $O(N)$ без создания массива на $\frac{B - A}{2}$ чисел.
6. ***random_arange_ndarray*** - пиковый расход - 3817.23 МБ, временная сложность: O($M$) для `np.arange` и O($N$) для выборки (`np.choice`). В настоящей версии numpy `rng.choice` больше не перемешивает весь массив, но затраты для `np.arange` остаются теми же

### 2. Преобразуйте каждый элемент массива по формуле Softmax. Найдите произведение элементов.

$softmax$:  $\displaystyle \sigma$ = $\displaystyle \frac{e^{x_i}}{\sum_{j=1}^{n} e^{x_j}}$

````python
import numpy as np

def softmax(x: np.ndarray) -> np.ndarray:
    '''
    Функиця для преобразования входного массива по формуле softmax
    Возвращает преобразованный массив
    '''
    exp = np.exp(x)
    return exp / np.sum(exp)

arr = np.array([1, 2, 3])
arr_soft = softmax(arr)
print(f'arr после softmax: {arr_soft}')
print(f'Произведение элементов = {np.prod(arr_soft)}')

#Проверка функции (аналог assert в numpy)
np.testing.assert_allclose(softmax(np.array([1, 1, 1])), np.array([1/3, 1/3, 1/3]))
np.testing.assert_allclose(softmax(np.array([0, 0, 0])), np.array([1/3, 1/3, 1/3]))
np.testing.assert_allclose(softmax(np.array([-1, -1, -1])), np.array([1/3, 1/3, 1/3]))
````
